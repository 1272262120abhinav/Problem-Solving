```python
#Print the 4th element from first and 4th element from last in a tuple

n = (10, 20, 30, 40, 50, 60, 70, 80, 90)
print("Tuple:",n)
print("4th element from first:",n[3])
print("4th element from last:",n[-4])
```

```python
#Check whether an element exists in a tuple or not

n = (10, 20, 30, 40, 50)
element = int(input("Enter element to search:"))
if element in n:
    print("Element exists")
else:
    print("Element does not exist")
```

```python
#Convert a list into a tuple

n_list = [10, 20, 30, 40, 50]
n_tuple = tuple(n_list)
print("List:", n_list)
print("Tuple:", n_tuple)
```

```python
#Find the index of an item in a tuple

n = (10, 20, 30, 40, 50)
element = int(input("Enter element:"))
if element in n:
    print("Index of the element is:", n.index(element))
```

```python
#WAP to replace last value of tuples in a list to 100

n = [(10, 20, 50), (50, 60, 80), (90, 100, 110)]
new_list = []
for tup in n:
    new_tup = tup[:-1] + (120,)
    new_list.append (new_tup)
print("Original list:", n)
print("Updates list:", new_list)
```

```python
#Find maximum and minimum element in a tuple

n = (10, 67, 89, 91, 1, 18752)
maximum = max(n)
minimum = min(n)
print("Tuple:", n)
print("Maximum element:", maximum)
print("Minimum element:", minimum)
```

```python
#Swap the first and last element of a tuple

n = (11, 22, 33, 44, 54)
new_tuple = (n[-1],) + n[1:-1] + (n[0],)
print("Original Tuple:", n)
print("Tuple after swapping:",new_tuple)
```
