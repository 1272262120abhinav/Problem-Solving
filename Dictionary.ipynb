```python
#WAP to add a key and value to dictionary

my_dict = {0 : 10, 1 : 20}
my_dict[2] = 30
print("Dictionary:", my_dict)
```

```python
#WAP to concatenate the following dictonaries to create a new one

dic1 = {1: 10, 2: 20}
dic2 = {3: 30, 4: 40}
dic3 = {5: 50, 6: 60}
new_dic = {}
new_dic.update(dic1)
new_dic.update(dic2)
new_dic.update(dic3)
print("New Dictionary:", new_dic)
```

```python
#WAP to check if a given key already exists in a dictionary

my_dict = {101:"Abc", 102:"Def", 103:"Ghi"}
key = int(input("Enter key to search:"))
if key in my_dict:
    print("Key exists in the dictionary")
else:
    print("Key does not exist in the dictionary")
```

```python
#WAP to iterate over a dictionary using for loop and print the keys alone, values alone and both keys and values

my_dict = {
    101: "Abc",
    102: "Def",
    103: "Ghi",
}
print("Keys:")
for key in my_dict:
    print(key)
print("\nValues:")
for value in my_dict.values():
    print(value)
print("\nKeys and Values:")
for key, value in my_dict.items():
    print(key, value)
```

```python
#WAP to prepare dictionary where the keys are numbers betn 1 and 15 (both included) and the values are square of the keys

my_dict = {}
for i in range(1, 16):
    my_dict[i] = i**2
    print(my_dict)
```

```python
#WAP to sum all values in a dictionary, considering the values will be of integer type

my_dict = { "a":10, "b":20, "c":30, "d":40 }
total = 0 
for value in my_dict.values():
    total = total + value
    print("Sum of all values:",total)
```
